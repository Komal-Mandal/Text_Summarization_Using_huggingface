This function splits a list into smaller batches. For example, if you have a list of 8 items and want batches of 3, it will create chunks like [1, 2, 3], [4, 5, 6], and [7, 8]. This helps in processing the list in smaller parts.

```python
def generate_batch_sized_chunks(list_of_elements, batch_size):

    for i in range(0, len(list_of_elements), batch_size):
        yield list_of_elements[i : i + batch_size]
```

Split the data into small batches of articles and summaries.

Tokenize the articles (turn them into a format the model can understand).

Generate summaries using the model for each article.

Decode the generated summaries back into readable text.

Compare the generated summaries to the real summaries and add them to the ROUGE metric.

Compute and return the ROUGE score to see how good the model's summaries are

```python
def calculate_metric_on_test_ds(dataset, metric, model, tokenizer,
                               batch_size=5, device=device,
                               column_text="article",
                               column_summary="highlights"):
    article_batches = list(generate_batch_sized_chunks(dataset[column_text], batch_size))
    target_batches = list(generate_batch_sized_chunks(dataset[column_summary], batch_size))

    for article_batch, target_batch in tqdm(
        zip(article_batches, target_batches), total=len(article_batches)):

        inputs = tokenizer(article_batch, max_length=128,  truncation=True,
                        padding="max_length", return_tensors="pt")

        summaries = model.generate(input_ids=inputs["input_ids"].to(device),
                         attention_mask=inputs["attention_mask"].to(device),
                         length_penalty=0.8, num_beams=5, max_length=50)
        ''' parameter for length penalty ensures that the model does not generate sequences that are too long. '''

        # Finally, we decode the generated texts,
        # replace the  token, and add the decoded texts with the references to the metric.
        decoded_summaries = [tokenizer.decode(s, skip_special_tokens=True,
                                clean_up_tokenization_spaces=True)
               for s in summaries]

        decoded_summaries = [d.replace("", " ") for d in decoded_summaries]


        metric.add_batch(predictions=decoded_summaries, references=target_batch)

    #  Finally compute and return the ROUGE scores.
    score = metric.compute()
    return score
```

*   convert_examples_to_features:

Input: Takes a batch of examples (dialogues and summaries).

Tokenization:
It tokenizes the dialogue (the text to summarize) using the tokenizer, truncating to a max length of 1024 tokens.

Then, it tokenizes the summary (the reference summary) as the target, truncating to a max length of 128 tokens.

* Output: Returns a dictionary containing:

1.input_ids:

These are the encoded representations of the dialogue (input text) that the model uses. The text is converted into numbers (tokens) that the model can process.

2.attention_mask:

This mask tells the model which parts of the input are actual text and which parts are just padding (empty space added for consistency). It helps the model ignore the padding when making predictions.

3.labels:

These are the encoded representations of the summary (target text) that the model should aim to generate. The model compares its predictions to these labels to learn and improve.


* dataset_samsum.map():

This applies the convert_examples_to_features function to the dataset_samsum.
batched=True: This processes multiple examples at once for **efficiency**






```python
def convert_examples_to_features(example_batch):
    input_encodings = tokenizer(example_batch['dialogue'] , max_length = 1024, truncation = True )

    with tokenizer.as_target_tokenizer():
        target_encodings = tokenizer(example_batch['summary'], max_length = 128, truncation = True )

    return {
        'input_ids' : input_encodings['input_ids'],
        'attention_mask': input_encodings['attention_mask'],
        'labels': target_encodings['input_ids']
    }

dataset_samsum_pt = dataset_samsum.map(convert_examples_to_features, batched = True)
```

This code creates a helper (data collator) that prepares the data for the model. It automatically handles:

Batching: Groups the data into batches.

Padding: Adds extra space to make all sequences the same length (so the model can process them together).

The seq2seq_data_collator is ready to be used when training or evaluating the model, ensuring that the input and output are properly formatted.








```python
from transformers import DataCollatorForSeq2Seq

seq2seq_data_collator = DataCollatorForSeq2Seq(tokenizer, model=model_pegasus)
```

This code sets up the settings for training the model, including:

Where to save the model (output_dir='pegasus-samsum').

How many times to train the model (num_train_epochs=1).

How many steps to warm up the learning rate (warmup_steps=500).

Batch size for training and evaluation (per_device_train_batch_size=1).

How often to log, evaluate, and save the model (logging_steps=10, eval_steps=500, save_steps=1e6).

Regularization to avoid overfitting (weight_decay=0.01).

Accumulate gradients for larger effective batches (gradient_accumulation_steps=16).

```python
from transformers import TrainingArguments, Trainer

trainer_args = TrainingArguments(
    output_dir='pegasus-samsum', num_train_epochs=1, warmup_steps=500,
    per_device_train_batch_size=1, per_device_eval_batch_size=1,
    weight_decay=0.01, logging_steps=10,
    evaluation_strategy='steps', eval_steps=500, save_steps=1e6,
    gradient_accumulation_steps=16
)
```

This code creates a Trainer to train the model with the following settings:

Model: It uses the model_pegasus for training.

Arguments: The trainer_args define how the training should be done (batch size, number of epochs, etc.).

Tokenizer: It uses the tokenizer to process the text.
Data collator: The seq2seq_data_collator handles how the data is batched and padded.

Training dataset: It uses dataset_samsum_pt["train"] for training.

Evaluation dataset: It uses dataset_samsum_pt["validation"] for evaluation.

In short, it sets up everything needed to train and evaluate the model using the provided datasets and configurations.








```python
trainer = Trainer(model=model_pegasus, args=trainer_args,
                  tokenizer=tokenizer, data_collator=seq2seq_data_collator,
                  train_dataset=dataset_samsum_pt["train"],
                  eval_dataset=dataset_samsum_pt["validation"])
```

this command runs the model through the training process using the data you provided, updating the model's weights to improve its performance.

```python
trainer.train()
```

This code checks how well the trained model summarizes text by comparing its summaries to the real ones in the test data. It calculates the ROUGE F1 scores (a measure of similarity) for different ROUGE types (rouge1, rouge2, rougeL, rougeLsum), then displays these scores in a table.

In short, it evaluates the model's performance and shows the results in a clear format.

```python
score = calculate_metric_on_test_ds(
    dataset_samsum['test'], rouge_metric, trainer.model, tokenizer, batch_size = 2, column_text = 'dialogue', column_summary= 'summary'
)

rouge_dict = dict((rn, score[rn] ) for rn in rouge_names )

pd.DataFrame(rouge_dict, index = [f'pegasus'] )
```
